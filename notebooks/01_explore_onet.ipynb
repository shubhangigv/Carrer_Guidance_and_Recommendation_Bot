# 01 · Explore O*NET 31.0

Read-only look at the raw O*NET text database before writing any recommender code.

Data: `data/raw/onet/` (download `db_31_0_text.zip` from https://www.onetcenter.org/db_releases.html and extract it there).

```python
import pandas as pd
from pathlib import Path

pd.set_option("display.max_rows", 100)

ONET_DIR = Path("..") / "data" / "raw" / "onet"
files = sorted(ONET_DIR.rglob("*.txt"))
print(len(files), "O*NET files found")
```

```python
# Every file with its number of rows and columns
rows = []
for f in files:
    try:
        df = pd.read_csv(f, sep="\t", dtype=str)
        rows.append({"file": f.name, "rows": len(df), "columns": len(df.columns)})
    except Exception as e:
        rows.append({"file": f.name, "rows": "ERROR", "columns": str(e)[:60]})

inventory = pd.DataFrame(rows)
inventory
```

```python
# The list of occupations
occ_file = next(f for f in files if f.name == "Occupation Data.txt")
occupations = pd.read_csv(occ_file, sep="\t", dtype=str)
print("Number of occupations:", len(occupations))
occupations.head(10)
```

Finding: O*NET 31.0 renamed some files. `Interests.txt` is now `Career Interest Types.txt`,
and `Skills.txt` is split into `Essential Skills.txt` (10) and `Transferable Skills.txt` (25).

```python
# All file names, then columns, scales and elements of the files the recommender uses
print("ALL FILES:")
print([f.name for f in files])

def find(name):
    return next((x for x in files if x.name == name), None)

for name in ["Career Interest Types.txt", "Essential Skills.txt",
             "Transferable Skills.txt", "Knowledge.txt", "Abilities.txt"]:
    f = find(name)
    if f is None:
        print("\n" + name, "-> NOT FOUND")
        continue
    df = pd.read_csv(f, sep="\t", dtype=str)
    print(f"\n{name}: {len(df)} rows")
    print("  Columns :", list(df.columns))
    if "Scale ID" in df.columns:
        print("  Scales  :", df["Scale ID"].unique().tolist())
    if "Element Name" in df.columns:
        print("  Elements:", df["Element Name"].unique().tolist())
```
